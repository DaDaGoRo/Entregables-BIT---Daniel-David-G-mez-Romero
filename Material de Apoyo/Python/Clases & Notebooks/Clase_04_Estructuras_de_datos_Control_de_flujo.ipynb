## Conversion de tipos de datos

### Enteros

````python
numero_string = "123" # Cadena de caracteres
numero_entero = int(numero_string)
print(numero_entero + 1)
print(f"Tipo de dato de la variable numero_string: {type(numero_string)}    str = cadena de caracteres, texto")
print(f"Tipo de dato de la variable numero_entero: {type(numero_entero)}    int = integer, numero entero")
````

### Decimales

````python
temperatura_decimal = 28.6
temperatura_entero = int(temperatura_decimal)
print(temperatura_entero)
print(f"Tipo de dato de la variable temperatura_decimal: {type(temperatura_decimal)}    float = numero con decimales")
print(f"Tipo de dato de la variable temperatura_entero: {type(temperatura_entero)}")
````

### Redondeo o aproximación de decimales a enteros

````python
# Ejemplo de aproximación antes de convertir a int
precio_original = 99.9  # Precio con decimales
precio_redondeado = round(precio_original)  # Redondear al valor entero más cercano
precio_entero = int(precio_original)  # Convertir a entero
print(f"Precio redondeado: {precio_redondeado}")
print(f"Precio entero: {precio_entero}")
````

### Tuplas (Inmutable - Longitud Fija)

````python
# Crear la tupla a partir de una secuencia de valores entre parentesis separados por comas
tupla = (4, 5, 6)
print(tupla)
````

````python
# Crear la tupla a partir de una secuencia de valores separados por comas
tupla_2 = 4, 5, 6 , "Luis"
print(tupla_2)
print(type(tupla_2))
````

````python
# Crear la tupla usando la funcion tuple sobre cualquier secuencia de datos
tupla_3 = tuple("hola")
tupla_3
````

````python
#Revisar si un objeto pertenece a la tupla
print(f"El número 10 está en la tupla_2? {5 in tupla_2}")
print(f"El número 50 está en la tupla_2? {50 in tupla_2}")
````

````python
#Acceder a los datos de la tupla
print(f"Primer elemento de la tupla: {tupla_2[0]}") #INDEXACION
print(f"Segundo elemento de la tupla: {tupla_2[1]}")
print(f"Tercer elemento de la tupla: {tupla_2[2]}")
````

### Diccionarios (Llave - Valor)

````python
datos_personales = {"id":1231412312,
                    "nombre": "Juana",
                    "edad": 10,
                    "equipo_favorito":"colombia"
                  }
datos_personales
````

````python
# Puedo leer los datos almacenados en una llave usando parentesis cuadrados
datos_personales["equipo_favorito"]
````

````python
# Tambien puedo leer los datos almacenados en una llave usando la funcion get()
datos_personales.get("equipo_favorito")
````

````python
# Para agregar nuevas llaves al diccionario, declaro la nueva llave entre parentesis cuadrados y le asigno un nuevo valor
datos_personales["fecha_nacimiento"] = "31/08/1998"
datos_personales["signozodiacal"] = "Sagitario"

datos_personales
````

````python
# Puedo borrar datos del diccionario usando la sentencia 'del' delete
del datos_personales['signozodiacal']
datos_personales
````

````python
# Tambien puedo borrar datos del diccionario usando la funcion pop() especificando la llave que quiero eliminar
datos_personales.pop("fecha_nacimiento")
datos_personales
````
