### Conceptos Clave:

- Series: Arreglo unidimensional etiquetado

- DataFrame: Estructura bidimensional tipo tabla

- Índices: Etiquetas para acceder a los datos

## Series vs Dataframe

```python
# Una serie es una columna

edades = pd.Series([20,25,30])

edades
```

```python
# Un Dataframe es una tabla

datos = {
    "Nombre":["Ana","Luis","Pedro"],
    "Edad":[20,25,30],
    "Ciudad":["Bogotá","Pereira","Cali"]
}

df = pd.DataFrame(datos)

df
```

## Filtrar información

```python
# Clientes de Pereira

clientes[clientes["ciudad"] == "Pereira"]
```

```python
clientes[clientes['nombre_cliente'].str.startswith("Lu")] # Startswith - Empezar con
```

```python
clientes[clientes['nombre_cliente'].str.contains("an")] # Contains - contiene
```

```python
clientes[clientes['nombre_cliente'].str.endswith("ez")] # Endswith - Termine en
```

```python
# Dos condiciones
clientes[
    (clientes["segmento"]== "Regular") & # & and-y
    (clientes["ciudad"]=="Manizales") | # | or-o
    (clientes["nombre_cliente"].str.startswith("Lu"))
]
```

```python
clientes_filtrados = [nombre_cliente for nombre_cliente in clientes["nombre_cliente"]
                      if "Daniel" in nombre_cliente]

clientes_filtrados
```
