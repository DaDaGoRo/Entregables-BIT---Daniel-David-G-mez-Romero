# 4. Tratamiento de valores nulos

Hay varias estrategias posibles.

### Opción A — Eliminar filas

```python
df.dropna()
```

### Opción B — Eliminar según una columna específica

```python
df.dropna(subset=["ciudad"])
```

### Opción C — Completar con un valor

```python
df["ciudad"].fillna("Desconocida")
```

### Opción D — Imputar con una medida

Por ejemplo, para una variable numérica:

```python
df["edad_cliente"].fillna(df["edad_cliente"].median())
```

### Decisión de negocio

No existe una función universalmente correcta.

La pregunta no es:

> "¿Qué función uso?"

La pregunta es:

> **"¿Qué debería significar un valor faltante en este contexto?"**


````python
df[df["ciudad"].isnull()].head(10)
````

````python
df["ciudad"] = df["ciudad"].fillna("Desconocida")
````

Para `edad_cliente`, en cambio, podemos utilizar la mediana como ejercicio de imputación.

¿Por qué la mediana y no la media?

Porque suele ser menos sensible a valores extremos.


````python
mediana_edad = df["edad_cliente"].median()

df["edad_cliente"] = df["edad_cliente"].fillna(mediana_edad)

mediana_edad
````

````python
df["edad_cliente"] = df["edad_cliente"].fillna(mediana_edad)
````

# 6. Limpieza de texto

Los más importantes:

- `.str.strip()` → elimina espacios al inicio y final
- `.str.lower()` → minúsculas
- `.str.upper()` → mayúsculas
- `.str.title()` → Iniciales en mayúscula
- `.str.replace()` → reemplazar texto


````python
df["ciudad"].value_counts()
````

````python
df["ciudad"] = (
    df["ciudad"]
    .str.strip()
    .str.lower()
    .str.title()
)

df["ciudad"].value_counts()
````

````python
df["ciudad"] = (
    df["ciudad"]
    .str.replace("Medellin", "Medellín", regex=False)
    .str.replace("Bogota", "Bogotá", regex=False)
)
````

````python
df["ciudad"].value_counts()
````

````python
df["categoria"] = (
    df["categoria"]
    .str.replace("Tecnología", "Tecnologia", regex=False)
)
````

# 7. Tipos de datos


````python
df.info()
````

### Problema

`precio_unitario` puede estar almacenado como texto porque contiene valores como:

```text
2500000
$850.000
```



````python
df["precio_unitario"].head(10)
````

## Limpiar el precio

Eliminar el símbolo `$` y los separadores de miles.


````python
df["precio_unitario"] = (
    df["precio_unitario"]
    .astype("string") # .astype () - convierte columna a texto
    .str.replace("$", "", regex=False)
    .str.replace(".", "", regex=False) # regex=False = busca exactamente ese texto, sin interpretarlo como una expresión regular
)
````

````python
df["precio_unitario"].head(10)
````

### Convertir a número

`pd.to_numeric()` intenta convertir los valores a un tipo numérico.

Con `errors="coerce"` los valores que no puedan convertirse pasan a ser `NaN`.


````python
df["precio_unitario"] = pd.to_numeric(
    df["precio_unitario"],
    errors="coerce"
)

df["precio_unitario"].dtype
````

# 8. Limpiar descuentos %

Los descuentos aparecen con formatos como:

```text
10%
5%
0%
0.1
```




````python
df["descuento"].head(10)
````

````python
df["descuento"] = (
    df["descuento"]
    .astype("string")
    .str.replace("%", "", regex=False)
)

df["descuento"] = pd.to_numeric(
    df["descuento"],
    errors="coerce"
)
````

Ahora tenemos los porcentajes como números.

Para trabajar con el descuento en cálculos, se puede convertir a proporción:

`10` → `0.10`

`5` → `0.05`


````python
df["descuento"] = df["descuento"] / 100
````

````python
df["descuento"].head(10)
````

# 10. Valores inválidos

Un dato puede tener el tipo correcto y aun así estar mal.

Ejemplo:

```text
edad_cliente = 150
cantidad = -3
```



## Revisión de las edades


````python
df[df["edad_cliente"] < 18][["id_venta", "cliente", "edad_cliente"]].head(20)
````

Se consideraran válidas las edades entre 18 y 100 años.



````python
df[
    (df["edad_cliente"] < 18) |
    (df["edad_cliente"] > 100)
][["id_venta", "cliente", "edad_cliente"]].head(20)
````

En lugar de borrar inmediatamente las filas, se convertiran esos valores inválidos en faltantes.


````python
condicion_edad = (
    (df["edad_cliente"] < 18) |
    (df["edad_cliente"] > 100)
)

df.loc[condicion_edad, "edad_cliente"] = pd.NA
````

## Revisamos las cantidades

Una venta no debería tener una cantidad menor o igual a cero.


````python
df[df["cantidad"] <= 0][
    ["id_venta", "producto", "cantidad"]
]
````

Se consideraran estos registros como inválidos y se eliminaran.

**Importante:** no siempre se eliminan datos inválidos. Esta es una decisión de negocio.


````python
df = df[df["cantidad"] > 0].copy()
````

# 13. Crear variables DESPUÉS de limpiar


**Subtotal = cantidad × precio unitario**


````python
df["subtotal"] = (
    df["cantidad"] *
    df["precio_unitario"]
)
````

Calcular el valor del descuento.


````python
df["valor_descuento"] = (
    df["subtotal"] *
    df["descuento"]
)
````

Calculo del total.


````python
df["total_venta"] = (
    df["subtotal"] -
    df["valor_descuento"]
)
````
